```python

import numpy as np
import pandas as pd

data = [[1, 'Smith', 50000],
        [2, 'Jones', 60000]]

df = pd.DataFrame(data)
df.columns = ['Empd', 'Name', 'Salary']

print("DataFrame Information:")
df.info()
print()
print("\nDescriptive Statistics:")
print(df.describe())
print()
print("\nFirst 5 Rows:")
print(df.head())
print()
print("\nLast 5 Rows:")
print(df.tail())
print()
print("\nEmployee Names:")
print(df['Name'])
print()
print("\nAverage Salary:")
print(df['Salary'].mean())
```
